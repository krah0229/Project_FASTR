## Notes for Task 3:

- Params is an optional argument that allows us to pass our own arguments into the event handling function through cv.setMouseCallback().
- cv.setMouseCallback() takes a reference to your mouse event function as an argument.

## Task 3: Drawing a line with mouse events

```python
import cv2 as cv
import numpy as np

# Globals to track drawing state
x0, y0 = None, None
drawing = False
shape = "line"  # Current shape mode — 'l' = line, 'r' = rectangle

def draw_shape(event, x, y, flags, *params):
    """Mouse event handler — draws a line or rectangle based on current shape mode."""
    global x0, y0, drawing, shape

    image, = params  # Unpack the params tuple
    colour = (0, 0, 255)  # Red in BGR

    # Start drawing on left mouse button down
    if event == cv.EVENT_LBUTTONDOWN:
        drawing = True
        x0, y0 = x, y

    # Finalise the shape when button released
    elif event == cv.EVENT_LBUTTONUP and drawing:
        drawing = False
        if shape == "line":
            cv.line(image, (x0, y0), (x, y), colour, 2)
        elif shape == "rectangle":
            cv.rectangle(image, (x0, y0), (x, y), colour, 2)

def main():
    global shape

    # Create a 720x720 blank black image with 3 colour channels
    blank = np.zeros((720, 720, 3), dtype=np.uint8)

    cv.namedWindow("Image")
    cv.setMouseCallback("Image", draw_shape, param=(blank,))

    print("Controls: L = line mode, R = rectangle mode, D = quit")

    while True:
        cv.imshow("Image", blank)
        key = cv.waitKey(1) & 0xFF

        if key == ord('d'):
            break
        elif key == ord('l'):
            shape = "line"
            print("Mode: line")
        elif key == ord('r'):
            shape = "rectangle"
            print("Mode: rectangle")

    cv.destroyAllWindows()

main()
```
