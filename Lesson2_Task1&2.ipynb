Task 1: Drawing Shapes
1. Create a new python file, importing OpenCV and Numpy.
2. Create a blank image using the Numpy function np.zeros([height, width, 3]). 3 specifies
the number of colour channels.
3. Using the functions above, draw one of each shape on the blank canvas with varying
colours.
4. Display the image using cv.imshow().

```python
import cv2 as cv
import numpy as np

# Create a blank black image (500x500 pixels, 3 colour channels BGR)
canvas = np.zeros([500, 500, 3], dtype=np.uint8)

# Draw a line from top-left to bottom-right (green, 3px thick)
cv.line(canvas, (0, 0), (500, 500), (0, 255, 0), 3)

# Draw a rectangle using two opposite corner points (red, 2px thick)
cv.rectangle(canvas, (100, 100), (400, 250), (0, 0, 255), 2)

# Draw a filled circle at centre (250, 350) with radius 50 (blue, filled with -1)
cv.circle(canvas, (300, 350), 100, (200, 120, 0), -1)

# Draw a polygon using an array of points (yellow, 2px thick)
points = np.array([[400, 100], [480, 200], [440, 300], [360, 300], [320, 200]], np.int32)
cv.polylines(canvas, [points], True, (0, 215, 255), 2)

cv.imshow("Canvas", canvas)
```

```python
import matplotlib.pyplot as plt

# Normal text, FONT_HERSHEY_SIMPLEX, scale 1, white, 2px thick
cv.putText(canvas, "Hello FASTR!", (150, 50),
           cv.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

# Different font and larger scale (orange)
cv.putText(canvas, "OpenCV", (150, 90),
           cv.FONT_HERSHEY_TRIPLEX, 1.5, (0, 165, 255), 3)

# Far left x=0 — text gets clipped at the canvas edge
cv.putText(canvas, "Far Left Text point", (0, 300),
           cv.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 0), 2)

# Display inline
plt.imshow(cv.cvtColor(canvas, cv.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
