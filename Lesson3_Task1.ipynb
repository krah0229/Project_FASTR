## Task 1: Creating an Intensity Threshold
1. Create a new python file and import the OpenCV library.
2. Find and download a video if you do not have access to a webcam.
3. Create a capture object with the video or using your webcam and start reading through
the video frames using a while True loop. Make sure to not set the delay argument of
cv.waitKey() too low.
4. Convert the frame to grayscale and apply a basic threshold as well as an adaptive
threshold and display them using cv.imshow(). What differences can you note between
the two?

```python
import cv2 as cv

capture = cv.VideoCapture("FASTR_Gandalf.mp4")

while True:
    retval, frame = capture.read()

    if retval == False:
        break

    # Convert frame to grayscale (required for thresholding)
    gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)

    # Basic threshold — pixels above 127 become white (255), below become black (0)
    _, basic_thresh = cv.threshold(gray, 127, 255, cv.THRESH_BINARY)

    # Adaptive threshold — compares each pixel to its local neighbourhood (block of 11x11)
    adaptive_thresh = cv.adaptiveThreshold(gray, 255,
                                           cv.ADAPTIVE_THRESH_GAUSSIAN_C,
                                           cv.THRESH_BINARY, 11, 2)

    # Display all three
    cv.imshow("Grayscale", gray)
    cv.imshow("Basic Threshold", basic_thresh)
    cv.imshow("Adaptive Threshold", adaptive_thresh)

    if cv.waitKey(1) & 0xFF == ord('d'):
        break

capture.release()
cv.destroyAllWindows()
```
