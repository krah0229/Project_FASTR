## Notes for Task 4:
- cv.blur(image: cv.Mat, kernel_size: tuple) -> Applies standard average blur to an image
- cv.GaussianBlur(image: cv.Mat, kernel_size: tuple, x_s, y_s) -> Weighted average blurring
- cv.medianBlur(image: cv.Mat, kernel_size: int) -> Takes the median value, and blurs it
- cv.bilateralFilter(image: cv.Mat, d: int, colour_s: int, space_s) -> Blurs but preserves object edges
- cv.filter2D(image: cv.Mat, ddepth: int, kernel: np.ndarray) -> Apply our own kernel to blur the image

## Task 4: Comparing Blurs
1. Create a new python file importing the OpenCV library and matplotlib.pyplot. Download
the image of the backyard on page 6 or find an image of your own.
2. Apply a regular blur, gaussian blur, median blur and bilateral blur to the image, assigning
them to their own variables. Make sure to use an appropriate kernel size.
3. Create a subplot with dimensions (3 x 2) using matplotlib, recalling the function
plt.imshow(). Display the original along side the blurred images making sure to use the
function plt.tight_layout() before using plt.show(). Remove the x and y ticks or each subplot
calling plt.xticks([ ]) and plt.yticks([ ]).

```python
import cv2 as cv
import matplotlib.pyplot as plt

img = cv.imread("FASTR_Garden.png")
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Print image size to see how big it is
print(f"Image size: {img.shape[1]} x {img.shape[0]} pixels")

# Much larger kernels for high-res images
blur = cv.blur(img, (71, 71))
gaussian = cv.GaussianBlur(img, (71, 71), 0)
median = cv.medianBlur(img, 71)
bilateral = cv.bilateralFilter(img, 25, 200, 200)

# Bigger figure so differences are easier to spot
plt.figure(figsize=(12, 14))

plt.subplot(3, 2, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.xticks([])
plt.yticks([])

plt.subplot(3, 2, 2)
plt.imshow(cv.cvtColor(blur, cv.COLOR_BGR2RGB))
plt.title("Average Blur (71x71)")
plt.xticks([])
plt.yticks([])

plt.subplot(3, 2, 3)
plt.imshow(cv.cvtColor(gaussian, cv.COLOR_BGR2RGB))
plt.title("Gaussian Blur (71x71)")
plt.xticks([])
plt.yticks([])

plt.subplot(3, 2, 4)
plt.imshow(cv.cvtColor(median, cv.COLOR_BGR2RGB))
plt.title("Median Blur (71)")
plt.xticks([])
plt.yticks([])

plt.subplot(3, 2, 5)
plt.imshow(cv.cvtColor(bilateral, cv.COLOR_BGR2RGB))
plt.title("Bilateral Filter")
plt.xticks([])
plt.yticks([])

plt.tight_layout()
plt.show()
```

## Task 5: Video Blurring
1. Choose one of the blurs from Task 4 and apply it to video frames. Don’t use matplotlib
for this task as it will be too slow, use cv2.imshow() instead.

```python
import cv2 as cv

capture = cv.VideoCapture("FASTR_Gandalf.mp4")

while True:
    retval, frame = capture.read()

    if retval == False:
        break

    # Apply Gaussian blur to each frame
    blurred = cv.GaussianBlur(frame, (25, 25), 0)

    # Show original and blurred side by side
    cv.imshow("Original", frame)
    cv.imshow("Gaussian Blur", blurred)

    if cv.waitKey(1) & 0xFF == ord('d'):
        break

capture.release()
cv.destroyAllWindows()
```
